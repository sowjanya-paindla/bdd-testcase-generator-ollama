## Cell 3 — BDD Generator Code

```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate

tc_llm = ChatOllama(
    base_url='http://localhost:11434',
    model='qwen2.5-coder',
    temperature=0.3,
)

def generate_test_cases(user_story: str) -> str:
    template = """You are a QA Automation Engineer.
Task Requirements:

Testing Framework: Use the Gherkin (Given/When/Then) format for BDD-style scenarios.

Coverage Depth: Include Happy Path (positive), Edge Cases (boundary values), and Negative Scenarios (error handling/security).

Validation Criteria: Ensure each test case validates specific Acceptance Criteria and addresses the user persona's intent.

Detailed Metadata: For each scenario, provide a clear Title, Pre-conditions, and Expected Result.

Usability & Performance: Include at least two scenarios focusing on UI/UX consistency and system responsiveness.

Output Format: Present the results in a structured Markdown table or organized list for direct integration into a test management tool.

User Story: {user_story}"""
    chain = PromptTemplate.from_template(template) | tc_llm
    return chain.invoke({'user_story': user_story}).content

print('✅ Generator ready!')
```

## Cell 4 — Run Generator

```python
user_story = """
As a finance user, I want to navigate to the Payment Traceability screen from the FMC tab so that I can trace payment details efficiently.

Acceptance Criteria:
1. User can access the FMC tab and select Payment Traceability.
2. Payment Traceability screen loads with search filters.
3. Page allows input of Account ID, Payment Gateway ID and time period for payment tracking.
"""

print('Generating BDD test cases...')
print('Please wait 30-60 seconds...\n')

output = generate_test_cases(user_story)
print(output)
```

## Cell 5 — Save Output

```python
import os
os.makedirs('outputs', exist_ok=True)

with open('outputs/bdd_tests.feature', 'w') as f:
    f.write(output)

print('✅ Saved to outputs/bdd_tests.feature')
```
